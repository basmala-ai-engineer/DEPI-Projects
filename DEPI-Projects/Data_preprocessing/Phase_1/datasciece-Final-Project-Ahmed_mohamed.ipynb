###  To fix this issue You will fill these values with mean for numeric and mode for categorical

```python
# trying to fill missing values in my datset using for loop 
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col] = df[col].fillna(df[col].mean())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#check the Nulls 
df.isnull().sum()
```

```python
# fill  dataset with missing values.
import numpy as np

other_mask = df['member_gender'] == 'Other'
df.loc[other_mask, 'member_gender'] = np.random.choice(
    ['Male', 'Female'], 
    size=other_mask.sum()
)
```

# <a id="1"> 1 ) . What is the distribution for User type?</a>

# <a id="8"> 8) . What the average duration of trip for the categories of user type?</a>

```python
# Distribution of Age by user types
plt.figure(figsize = (8,5), dpi = 100)

sb.boxplot(data=df[df['duration_min'] <= 60], x='user_type', y='duration_min', color=sb.color_palette()[0])

plt.xlabel('User Type')
plt.ylabel('Trip Duration (minutes)')
plt.title('Trip Duration by User Type')
plt.show()

df.groupby('user_type', observed=True)['duration_min'].mean()
```

# Make visualization to easy get insight


> ### <mark>Your analsys and exploration here </mark>

- `Customers` have a noticeably higher average trip duration (about 23.9 minutes) than `Subscribers` (about 10.7 minutes).
- This makes sense: subscribers likely use the bikes for short daily commutes, while customers tend to rent bikes for longer, leisure-style trips.
